### Projeto Semantix
Analisar as causas e acontecimentos no trânsito do mês de agosto de 2026 
tendo como objetivo trazer padrões e soluções durante os períodos.

# Dados Usados
- Fonte: Portal de Dados Abertos do Estado de São Paulo.
- Link: https://dadosabertos.sp.gov.br/dataset/sinistros-infosiga

# Ferramentas Usadas
- Python
- Pandas
- Looker Studio
- PyCharm

# Etapas
- Importação dos dados.
- Limpeza dos Dados.
- Análise dos registros.
- Criação dos arquivos em CSV.
- Dashboard interativo é explicativo no Looker Studio.
